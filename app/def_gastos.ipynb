```python
dane = pd.read_excel("C:\\PEPE_TERRITORIAL\\data\\mun.xlsx")
```

```python
print(dane.columns.tolist())
```

```python
dane = dane.rename(columns={
    "Código": "Codigo_Dep",
    "Nombre": "Nombre_Dep",
    "Código.1": "Codigo_Mun",
    "Nombre.1": "Nombre_Mun",
    "Municipio\nIsla\nÁrea no municipalizada": "Tipo_DANE"
})
```

```python
dane = dane.rename(columns={
    " Código ": "Codigo_Mun",
    " Nombre ": "Nombre_Mun"
})
```

```python
print(dane.columns.tolist())
```

```python
dane["Codigo_Mun"] = (
    dane["Codigo_Mun"]
    .astype(str)
    .str.zfill(5)
)

df["Código DANE"] = (
    df["Código DANE"]
    .astype(str)
    .str.zfill(5)
)
```

```python
dane[["Codigo_Mun", "Nombre_Mun"]].head()
```

```python
df = df.merge(
    dane[["Codigo_Mun", "Nombre_Mun"]],
    left_on="Código DANE",
    right_on="Codigo_Mun",
    how="left"
)
```

```python
## Ver nulos
print(df["Nombre_Mun"].isna().sum())

## Ver filas con nulos
df[df["Nombre_Mun"].isna()][
    ["Código DANE", "Departamento", "Tipo de Entidad"]
].drop_duplicates().head(20)
```

```python
df[df["Nombre_Mun"].isna()][
    ["Código DANE", "Tipo de Entidad", "Departamento"]
].drop_duplicates().head(50)
```

```python
df["Nombre_Mun"] = df["Nombre_Mun"].fillna(df["Departamento"])
```
